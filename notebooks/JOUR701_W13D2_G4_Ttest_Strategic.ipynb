## 2 · Prepare the data

We will remove repeated participant IDs and exclude missing values only from the columns used in this analysis. This is the same cleaning logic practiced in Week 10.

```python
# One record per participant. Keep only rows with both analysis variables.
df = raw.drop_duplicates(subset="participant_id").dropna(subset=["message_style", "share_intention"]).copy()
print("Usable participants:", len(df))
print("Rows removed:", len(raw) - len(df))
print("\nVariable summary:")
display(df[["message_style", "share_intention"]].describe(include="all"))
```

## 3 · Research question and hypothesis

**RQ:** Is average intention to share a public health message different between a facts message and a personal-story message?

**H₀: The two message styles have the same population mean sharing intention.**

Two independent randomized groups and one numeric outcome call for an independent-samples **t-test**.

## 4 · Compare the group means and standard deviations

```python
summary = df.groupby("message_style")["share_intention"].agg(["count", "mean", "std"])
summary.columns = ["N", "Mean", "SD"]
display(summary.round(2))

df.boxplot(column="share_intention", by="message_style")
plt.title("Intention to Share by Message Style")
plt.suptitle("")
plt.xlabel("Message style")
plt.ylabel("Sharing intention (1–7)")
plt.show()
```

**YOUR ANSWER — Description:** The group with the higher sample mean is **_____** (mean **_____**). The other group has mean **_____**.

## 5 · Run the t-test

Read group means, SDs, and **p** first. The t statistic is reported too but is not the main interpretation for this lab.

```python
from scipy.stats import ttest_ind

facts = df.loc[df["message_style"] == "Facts", "share_intention"]
story = df.loc[df["message_style"] == "Personal story", "share_intention"]
t, p = ttest_ind(facts, story, equal_var=False)
print("t:", round(t, 3))
print("p-value:", round(p, 4))
print("Reject H0?", "Yes" if p < .05 else "No: fail to reject H0")
```

**YOUR ANSWER — Results:** t = **_____**, p = **_____**. We **reject / fail to reject** H₀.
