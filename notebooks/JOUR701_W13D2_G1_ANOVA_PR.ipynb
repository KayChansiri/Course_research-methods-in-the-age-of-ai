## 2 · Prepare the data

We will remove repeated participant IDs and exclude missing values only from the columns used in this analysis. This is the same cleaning logic practiced in Week 10.

```python
# One record per participant. Keep only rows with both analysis variables.
df = raw.drop_duplicates(subset="participant_id").dropna(subset=["response_style", "brand_trust"]).copy()
print("Usable participants:", len(df))
print("Rows removed:", len(raw) - len(df))
print("\nVariable summary:")
display(df[["response_style", "brand_trust"]].describe(include="all"))
```

## 3 · Research question and hypothesis

**RQ:** Does average brand trust differ across three crisis-response styles?

**H₀: Average brand trust is equal across apology, explanation, and denial messages.**

The predictor has **three groups**; the outcome is a **numeric score**. This calls for **one-way ANOVA**.

## 4 · Compare the group means first

```python
summary = df.groupby("response_style")["brand_trust"].agg(["count", "mean", "std"])
summary.columns = ["N", "Mean", "SD"]
display(summary.round(2))

# Visualize the distribution of scores by response style.
df.boxplot(column="brand_trust", by="response_style")
plt.title("Brand Trust by Crisis-Response Style")
plt.suptitle("")
plt.xlabel("Response style")
plt.ylabel("Brand trust (1–7)")
plt.show()
```

**YOUR ANSWER — Description:** The highest average trust is for **_____** (mean **_____**). The lowest average is for **_____** (mean **_____**).

## 5 · Run one-way ANOVA

Read the **F** statistic and the **PR(>F)** p-value. ANOVA tells us whether at least one population group mean differs; it does not by itself tell us which pairs differ.

```python
import statsmodels.formula.api as smf
import statsmodels.api as sm

model = smf.ols("brand_trust ~ C(response_style)", data=df).fit()
anova_table = sm.stats.anova_lm(model, typ=2)
display(anova_table.round(4))

p = float(anova_table.loc["C(response_style)", "PR(>F)"])
print("ANOVA p-value:", round(p, 4))
print("Reject H0?", "Yes" if p < .05 else "No: fail to reject H0")
```

## 6 · Which pairs differ? (Tukey follow-up)

This comparison is the same follow-up introduced in the ANOVA lesson. Focus on the adjusted p-values and group labels.

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd

tukey = pairwise_tukeyhsd(endog=df["brand_trust"], groups=df["response_style"], alpha=0.05)
print(tukey)
```
