### A quick guide to Python data types

- `int64` = whole numbers
- `float64` = numeric values that can contain decimals
- `object` = usually text / category labels

**But:** Python's data type is not automatically the variable's research role.

For example, `participant_id` is stored as a number, but it is an **identifier**, not a meaningful numeric measure.


# Step 3 — Check missing data

`isna()` asks whether a value is missing.  
`sum()` counts the missing values in each column.


```python
df.isna().sum()
```

For today's demonstration, we will remove rows that are missing our main outcome, `message_trust`.

This changes `df` in the current Colab session. It does **not** change the original CSV on GitHub.


```python
df = df.dropna(subset=["message_trust"])
```

```python
df.isna().sum()
```

# Step 4 — Look for impossible values

An unusual value is not automatically wrong.

For example:

- 8 hours/day of social media use = unusual, but possible
- 99 hours/day = impossible

`describe()` gives us a quick numerical summary.


```python
df["daily_social_media_hours"].describe()
```

# Step 5 — Check variable types and category labels


```python
df.dtypes
```

For categorical variables, `value_counts()` lets us see the categories Python currently sees.


```python
df["condition"].value_counts()
```

```python
df["primary_platform"].value_counts()
```

Look carefully for labels that mean the same thing but are written differently, such as:

- `Personalized` vs. `personalized`
- `Instagram` vs. `instagram`

We can standardize capitalization:


```python
df["condition"] = df["condition"].str.title()
```

```python
df["primary_platform"] = df["primary_platform"].str.title()
```

We can also explicitly ask pandas to treat `age` as numeric.  
`errors="coerce"` means that values pandas cannot interpret as numbers become missing (`NaN`) rather than crashing the code.


```python
df["age"] = pd.to_numeric(df["age"], errors="coerce")
```

Check the categories again after cleaning:


```python
df["condition"].value_counts()
```

```python
df["primary_platform"].value_counts()
```

# Step 6 — Does the value fit the measurement?

Suppose `share_intention` was measured on a **1–7 scale**.

First, look at the minimum and maximum.


```python
df["share_intention"].min()
```

```python
df["share_intention"].max()
```

If the maximum is greater than 7, inspect the invalid row:


```python
df[df["share_intention"] > 7]
```

The important idea is methodological, not just computational:

**Python may accept a number even when that number is impossible for the measure you used.**


# Step 7 — Describe the data


```python
df.describe()
```

For a categorical variable, count how many observations are in each group:


```python
df["condition"].value_counts()
```

Now calculate the average `message_trust` separately for the two experimental conditions:


```python
df.groupby("condition")["message_trust"].mean()
```

A difference in group means is a **descriptive pattern**.

It does not yet tell us whether the difference is statistically convincing. Later in Block 3, we will learn how a **t-test** addresses that question.


# Step 8 — Visual checks

Plots help us notice potential problems before analysis.

For now, you only need to recognize what the plot helps you inspect. You do **not** need to memorize specific statistical assumptions yet.


```python
df["message_trust"].hist()
```

### Histogram

Look for:

- the overall shape
- strong skew
- unusual or extreme values


```python
df.boxplot(column="message_trust", by="condition")
```

### Boxplot

Look for:

- the typical values in each group
- the spread of the groups
- possible outliers
- whether the group spreads look dramatically different

A plot does not **prove** an assumption. It helps you inspect the data.
