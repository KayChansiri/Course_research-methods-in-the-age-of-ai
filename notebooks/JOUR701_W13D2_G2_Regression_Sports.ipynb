## 2 · Prepare the data

We will remove repeated participant IDs and exclude missing values only from the columns used in this analysis. This is the same cleaning logic practiced in Week 10.

```python
# One record per participant. Keep only rows with both analysis variables.
df = raw.drop_duplicates(subset="participant_id").dropna(subset=["transparency_score", "story_credibility"]).copy()
print("Usable participants:", len(df))
print("Rows removed:", len(raw) - len(df))
print("\nVariable summary:")
display(df[["transparency_score", "story_credibility"]].describe(include="all"))
```

## 3 · Research question and hypothesis

**RQ:** How does perceived source transparency predict sports story credibility?

**H₀: The regression slope for transparency is 0.**

Both variables are numeric. In **simple linear regression**, the outcome is Y (`story_credibility`) and the predictor is X (`transparency_score`).

## 4 · Look at the plot and descriptive statistics

```python
display(df[["transparency_score", "story_credibility"]].describe().loc[["mean", "std", "min", "max"]].round(2))

plt.scatter(df["transparency_score"], df["story_credibility"], alpha=0.65)
plt.xlabel("Perceived transparency score (1–7)")
plt.ylabel("Story credibility score (1–7)")
plt.title("Source Transparency and Credibility")
plt.show()
```

**YOUR ANSWER — Describe the pattern:** As transparency increases, credibility seems to **increase / decrease / show no clear pattern**. One thing we notice about the plot is **_____**.

## 5 · Run the regression

Read **intercept**, **slope/coefficient**, **p-value**, and **R²**. An association in a survey does not by itself demonstrate causation.

```python
import statsmodels.formula.api as smf

model = smf.ols("story_credibility ~ transparency_score", data=df).fit()
print("Intercept:", round(model.params["Intercept"], 3))
print("Slope:", round(model.params["transparency_score"], 3))
print("Slope p-value:", round(model.pvalues["transparency_score"], 4))
print("R-squared:", round(model.rsquared, 3))
print("95% CI for slope:", model.conf_int().loc["transparency_score"].round(3).tolist())
```

## 6 · Add the best-fitting line

```python
import numpy as np

x = df["transparency_score"]
y = df["story_credibility"]
slope, intercept = np.polyfit(x, y, 1)
x_line = np.linspace(x.min(), x.max(), 100)
plt.scatter(x, y, alpha=0.55)
plt.plot(x_line, intercept + slope * x_line, color="steelblue", linewidth=2)
plt.xlabel("Transparency score")
plt.ylabel("Credibility score")
plt.title("Regression: Transparency Predicting Credibility")
plt.show()
```
