## 2 · Prepare the data

We will remove repeated participant IDs and exclude missing values only from the columns used in this analysis. This is the same cleaning logic practiced in Week 10.

```python
# One record per participant. Keep only rows with both analysis variables.
df = raw.drop_duplicates(subset="participant_id").dropna(subset=["ad_relevance", "purchase_intention"]).copy()
print("Usable participants:", len(df))
print("Rows removed:", len(raw) - len(df))
print("\nVariable summary:")
display(df[["ad_relevance", "purchase_intention"]].describe(include="all"))
```

## 3 · Research question and hypothesis

**RQ:** Are perceived ad relevance and purchase intention related?

**H₀: Pearson’s population correlation between the two scores is 0.**

Both measures are numeric. Use Pearson correlation to summarize their **linear relationship**.

## 4 · Describe and plot the scores

```python
display(df[["ad_relevance", "purchase_intention"]].describe().loc[["mean", "std", "min", "max"]].round(2))

plt.scatter(df["ad_relevance"], df["purchase_intention"], alpha=0.65)
plt.xlabel("Ad relevance rating (1–7)")
plt.ylabel("Purchase intention rating (1–7)")
plt.title("Ad Relevance and Purchase Intention")
plt.show()
```

**YOUR ANSWER — Description:** The dots look **positively / negatively / not clearly** related. One unusual observation (if any) is **_____**.

## 5 · Run Pearson correlation

**r** describes direction and strength. **p** evaluates evidence against the null correlation of zero.

```python
from scipy.stats import pearsonr

r, p = pearsonr(df["ad_relevance"], df["purchase_intention"])
print("Pearson r:", round(r, 3))
print("p-value:", round(p, 4))
print("Reject H0?", "Yes" if p < .05 else "No: fail to reject H0")
```

**YOUR ANSWER — Results:** r = **_____**, p = **_____**. The direction is **positive / negative** and the relationship appears **weak / moderate / strong**. We **reject / fail to reject** H₀.
