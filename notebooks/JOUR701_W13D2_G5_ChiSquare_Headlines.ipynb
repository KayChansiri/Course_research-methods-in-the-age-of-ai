## 2 · Prepare the data

We will remove repeated participant IDs and exclude missing values only from the columns used in this analysis. This is the same cleaning logic practiced in Week 10.

```python
# One record per participant. Keep only rows with both analysis variables.
df = raw.drop_duplicates(subset="participant_id").dropna(subset=["headline_style", "clicked_article"]).copy()
print("Usable participants:", len(df))
print("Rows removed:", len(raw) - len(df))
print("\nVariable summary:")
display(df[["headline_style", "clicked_article"]].describe(include="all"))
```

## 3 · Research question and hypothesis

**RQ:** Is article clicking (Yes/No) associated with headline style (Neutral/Sensational)?

**H₀: Headline style and clicking are independent.**

Both variables are categorical. Use the **chi-square test of independence** to compare observed with expected counts.

## 4 · Count the Yes and No responses

```python
counts = pd.crosstab(df["headline_style"], df["clicked_article"])
counts = counts.reindex(index=["Neutral", "Sensational"], columns=["No", "Yes"], fill_value=0)
print("Observed counts:")
display(counts)
print("Percentages within each headline group:")
row_pct = counts.div(counts.sum(axis=1), axis=0) * 100
display(row_pct.round(1))

row_pct["Yes"].plot.bar(color="steelblue", ylim=(0,100))
plt.ylabel("Clicked Yes (%)")
plt.xlabel("Headline style")
plt.title("Click Rate by Headline Style")
plt.xticks(rotation=0)
plt.show()
```

**YOUR ANSWER — Description:** The group with the higher observed click rate is **_____**. The two click rates are **_____ %** and **_____ %**.

## 5 · Expected counts and chi-square

**Expected = (row total × column total) ÷ grand total**. Python calculates the expected table and the chi-square test.

```python
from scipy.stats import chi2_contingency

# Example: expected Neutral + Yes count using the totals.
neutral_total = counts.loc["Neutral"].sum()
yes_total = counts["Yes"].sum()
grand_total = counts.to_numpy().sum()
expected_neutral_yes = neutral_total * yes_total / grand_total
print("Expected Neutral + Yes =", round(expected_neutral_yes, 2))

chi2, p, dof, expected = chi2_contingency(counts, correction=False)
expected_table = pd.DataFrame(expected, index=counts.index, columns=counts.columns)
print("Expected counts if there were no relationship:")
display(expected_table.round(2))
print("Chi-square (χ²):", round(chi2, 3))
print("Degrees of freedom:", dof)
print("p-value:", round(p, 4))
print("Smallest expected count:", round(expected.min(), 2))
print("Reject H0?", "Yes" if p < .05 else "No: fail to reject H0")
```
