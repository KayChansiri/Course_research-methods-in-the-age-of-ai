## Cleaning decision guide for today

- **Missing value:** first ask *which variable is missing and does this analysis need it?* For today's exercise, remove a row only when the **main outcome** is missing.
- **Inconsistent label:** standardize labels when they clearly mean the same category.
- **Impossible/invalid value:** verify against the questionnaire/codebook/source. If the true value cannot be recovered, treat it as missing rather than inventing a replacement.
- **Unusual but possible value:** **do not automatically delete it.** Check whether it could be real and document the decision.
- **True duplicate:** remove the duplicate copy so one participant is not counted twice.

Cleaning is not about making results look better. Every change needs a reason.
